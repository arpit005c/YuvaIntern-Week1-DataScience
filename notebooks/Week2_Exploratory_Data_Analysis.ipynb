## Project: Air Quality Analysis Using Python

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")
plt.rcParams["figure.figsize"] = (10, 6)

print("Libraries imported successfully!")
```

```python
df = pd.read_csv("AirQuality_cleaned.csv")

print("Dataset loaded successfully!")
print("Dataset shape:", df.shape)

display(df.head())
```

```python
print("Dataset Information")
print("===================")

df.info()

print("\nDescriptive Statistics")
display(df.describe().T)

print("\nMissing Values")
display(df.isnull().sum().sort_values(ascending=False))
```

```python

# Step 5: Distribution Analysis using Histograms

numeric_df = df.select_dtypes(include=np.number)

numeric_df.hist(
    figsize=(18, 14),
    bins=30,
    edgecolor="black"
)

plt.suptitle(
    "Distribution of Numerical Features in Air Quality Dataset",
    fontsize=16,
    y=1.02
)

plt.tight_layout()
plt.show()
```

```python

# Step 6: Outlier Analysis Using Boxplots

numeric_df = df.select_dtypes(include=np.number)

plt.figure(figsize=(16, 7))

sns.boxplot(data=numeric_df)

plt.title("Boxplot of Numerical Features")
plt.xlabel("Features")
plt.ylabel("Values")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

## Step 7: Correlation Analysis

Correlation analysis helps identify the strength and direction of
relationships between numerical features in the air quality dataset.
A positive correlation means two variables tend to increase together,
while a negative correlation means one tends to decrease as the other increases.
Correlation does not establish causation.

```python

# Step 7: Calculate and inspect correlations

numeric_df = df.select_dtypes(include=np.number)

correlation_matrix = numeric_df.corr()

# Display the strongest positive and negative feature relationships
correlation_pairs = correlation_matrix.where(
    np.triu(np.ones(correlation_matrix.shape), k=1).astype(bool)
).stack().sort_values(ascending=False)

print("Top 10 strongest positive correlations:")
display(correlation_pairs.head(10).to_frame("Correlation"))

print("\nTop 10 strongest negative correlations:")
display(correlation_pairs.tail(10).sort_values().to_frame("Correlation"))
```

```python

# Step 8: CO concentration over time

df["Datetime"] = pd.to_datetime(df["Datetime"], errors="coerce")

time_df = df.dropna(subset=["Datetime"]).sort_values("Datetime")

plt.figure(figsize=(14, 6))

plt.plot(
    time_df["Datetime"],
    time_df["CO(GT)"],
    linewidth=0.8
)

plt.title("Carbon Monoxide Concentration Over Time")
plt.xlabel("Date")
plt.ylabel("CO Concentration")
plt.tight_layout()
plt.show()
```

```python

# Step 9: Daily pollutant trend analysis

pollutants = ["CO(GT)", "NOx(GT)", "NO2(GT)"]

daily_pollutants = (
    df.set_index("Datetime")[pollutants]
      .resample("D")
      .mean()
)

plt.figure(figsize=(14, 6))

for column in pollutants:
    plt.plot(
        daily_pollutants.index,
        daily_pollutants[column],
        label=column
    )

plt.title("Daily Average Trends of CO, NOx and NO2")
plt.xlabel("Date")
plt.ylabel("Average Concentration")
plt.legend()
plt.tight_layout()
plt.show()
```

```python

# Step 10: Relationship between Temperature and CO

plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=df,
    x="T",
    y="CO(GT)",
    alpha=0.4
)

plt.title("Relationship Between Temperature and CO Concentration")
plt.xlabel("Temperature")
plt.ylabel("CO Concentration")
plt.tight_layout()
plt.show()
```

```python

# Calculate correlation between temperature and pollutants

weather_pollution_corr = df[
    ["T", "RH", "AH", "CO(GT)", "NOx(GT)", "NO2(GT)"]
].corr()

display(
    weather_pollution_corr.loc[
        ["T", "RH", "AH"],
        ["CO(GT)", "NOx(GT)", "NO2(GT)"]
    ].round(3)
)
```

```python

# Step 12: Monthly Average CO Concentration

df["Datetime"] = pd.to_datetime(df["Datetime"], errors="coerce")

monthly_co = (
    df.dropna(subset=["Datetime"])
      .set_index("Datetime")["CO(GT)"]
      .resample("MS")
      .mean()
)

plt.figure(figsize=(12, 6))

plt.bar(
    monthly_co.index.strftime("%b %Y"),
    monthly_co.values
)

plt.title("Monthly Average CO Concentration")
plt.xlabel("Month")
plt.ylabel("Average CO Concentration")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

display(monthly_co.to_frame("Average CO"))
```

```python

# Step 13: Compare pollutant distributions

pollutants = ["CO(GT)", "NOx(GT)", "NO2(GT)"]

pollutant_data = df[pollutants].copy()

# Standardize each feature using its mean and standard deviation
pollutant_scaled = (
    pollutant_data - pollutant_data.mean()
) / pollutant_data.std()

plt.figure(figsize=(10, 6))

sns.boxplot(data=pollutant_scaled)

plt.title("Standardized Distribution of Air Pollutants")
plt.xlabel("Pollutant")
plt.ylabel("Standardized Value (Z-score)")
plt.tight_layout()
plt.show()
```

```python

# Step 14: Export the analysis results

correlation_matrix.to_csv("week2_correlation_matrix.csv")
monthly_co.to_csv("week2_monthly_co_summary.csv")

print("Analysis results exported successfully.")
```

```python

# Step 15: Daily CO trend analysis

df["Datetime"] = pd.to_datetime(df["Datetime"], errors="coerce")

daily_co = (
    df.dropna(subset=["Datetime"])
      .set_index("Datetime")["CO(GT)"]
      .resample("D")
      .mean()
      .dropna()
)

plt.figure(figsize=(14, 6))

plt.plot(daily_co.index, daily_co.values, linewidth=1.2)

plt.title("Daily Average Carbon Monoxide Concentration")
plt.xlabel("Date")
plt.ylabel("Average CO Concentration")
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()

print("Number of days analyzed:", len(daily_co))
print("Average daily CO:", round(daily_co.mean(), 3))
print("Highest daily CO:", round(daily_co.max(), 3))
print("Lowest daily CO:", round(daily_co.min(), 3))
```

```python

# Step 16: Monthly pollution trend analysis

df["Datetime"] = pd.to_datetime(df["Datetime"], errors="coerce")

# Treat -200 as an invalid reading for the selected pollutants
pollutants = ["CO(GT)", "NOx(GT)", "NO2(GT)"]

monthly_df = df.copy()

for col in pollutants:
    monthly_df[col] = monthly_df[col].replace(-200, np.nan)

monthly_df = monthly_df.dropna(subset=["Datetime"])
monthly_df["Month"] = monthly_df["Datetime"].dt.to_period("M")

monthly_summary = (
    monthly_df.groupby("Month")[pollutants]
    .mean()
)

display(monthly_summary.round(3))

# Plot monthly averages
monthly_summary.plot(
    kind="line",
    figsize=(14, 6),
    marker="o"
)

plt.title("Monthly Average Air Pollutant Concentrations")
plt.xlabel("Month")
plt.ylabel("Average Concentration")
plt.xticks(rotation=45)
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()

# Export the results
monthly_summary.to_csv("week2_monthly_pollution_trends.csv")

print("Monthly pollution analysis exported successfully.")
```

```python

# Step 17: Hourly pollution pattern analysis

df["Datetime"] = pd.to_datetime(df["Datetime"], errors="coerce")

hourly_df = df.copy()

pollutants = ["CO(GT)", "NOx(GT)", "NO2(GT)"]

# Replace invalid sensor readings
for col in pollutants:
    hourly_df[col] = hourly_df[col].replace(-200, np.nan)

hourly_df = hourly_df.dropna(subset=["Datetime"])
hourly_df["Hour"] = hourly_df["Datetime"].dt.hour

hourly_summary = hourly_df.groupby("Hour")[pollutants].mean()

display(hourly_summary.round(3))

hourly_summary.plot(
    figsize=(12, 6),
    marker="o"
)

plt.title("Average Air Pollutant Concentrations by Hour")
plt.xlabel("Hour of Day (0–23)")
plt.ylabel("Average Concentration")
plt.xticks(range(24))
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()

hourly_summary.to_csv("week2_hourly_pollution_analysis.csv")

print("Hourly pollution analysis exported successfully.")
```

```python

# Step 18: Pollution patterns by day of the week

df["Datetime"] = pd.to_datetime(df["Datetime"], errors="coerce")

day_df = df.copy()

pollutants = ["CO(GT)", "NOx(GT)", "NO2(GT)"]

for col in pollutants:
    day_df[col] = day_df[col].replace(-200, np.nan)

day_df = day_df.dropna(subset=["Datetime"])

day_df["Day"] = day_df["Datetime"].dt.day_name()

# Keep days in calendar order
day_order = [
    "Monday", "Tuesday", "Wednesday", "Thursday",
    "Friday", "Saturday", "Sunday"
]

day_summary = (
    day_df.groupby("Day")[pollutants]
    .mean()
    .reindex(day_order)
)

display(day_summary.round(3))

day_summary.plot(
    kind="bar",
    figsize=(12, 6)
)

plt.title("Average Air Pollutant Concentrations by Day of Week")
plt.xlabel("Day of Week")
plt.ylabel("Average Concentration")
plt.xticks(rotation=45)
plt.grid(axis="y", alpha=0.3)
plt.tight_layout()
plt.show()

day_summary.to_csv("week2_day_of_week_pollution.csv")

print("Day-of-week analysis exported successfully.")
```

```python

# Step 19: Weekday vs weekend pollution comparison

df["Datetime"] = pd.to_datetime(df["Datetime"], errors="coerce")

comparison_df = df.copy()

pollutants = ["CO(GT)", "NOx(GT)", "NO2(GT)"]

# Exclude invalid pollutant readings
for col in pollutants:
    comparison_df[col] = comparison_df[col].replace(-200, np.nan)

comparison_df = comparison_df.dropna(subset=["Datetime"])

comparison_df["Day_Type"] = np.where(
    comparison_df["Datetime"].dt.dayofweek < 5,
    "Weekday",
    "Weekend"
)

comparison = comparison_df.groupby("Day_Type")[pollutants].mean()

display(comparison.round(3))

# Plot each pollutant separately for a clearer comparison
fig, axes = plt.subplots(1, 3, figsize=(14, 5))

for ax, pollutant in zip(axes, pollutants):
    comparison[pollutant].plot(
        kind="bar", ax=ax, rot=0
    )
    ax.set_title(pollutant)
    ax.set_xlabel("Day Type")
    ax.set_ylabel("Average Concentration")
    ax.grid(axis="y", alpha=0.3)

plt.suptitle("Weekday vs Weekend Pollution Levels")
plt.tight_layout()
plt.show()

comparison.to_csv("week2_weekday_weekend_comparison.csv")

print("Weekday/weekend comparison exported successfully.")
```

```python

# Step 20: Analyze strongest correlations

numeric_df = df.select_dtypes(include=np.number).copy()

# Exclude invalid -200 readings from the correlation analysis
numeric_df = numeric_df.replace(-200, np.nan)

correlation_matrix = numeric_df.corr()

# Extract unique feature pairs
upper_triangle = correlation_matrix.where(
    np.triu(np.ones(correlation_matrix.shape), k=1).astype(bool)
)

correlation_pairs = upper_triangle.stack().sort_values(
    ascending=False
)

print("Top 10 positive correlations:")
display(correlation_pairs.head(10).to_frame("Correlation").round(3))

print("Top 10 negative correlations:")
display(correlation_pairs.tail(10).sort_values().to_frame("Correlation").round(3))

# Visualize the correlation matrix
plt.figure(figsize=(12, 9))

sns.heatmap(
    correlation_matrix,
    cmap="coolwarm",
    center=0,
    annot=True,
    fmt=".2f",
    square=True
)

plt.title("Correlation Heatmap After Invalid-Value Treatment")
plt.tight_layout()
plt.show()

correlation_matrix.to_csv("week2_cleaned_correlation_matrix.csv")

print("Cleaned correlation analysis exported successfully.")
```

```python

# Step 21: Final EDA summary

print("=" * 55)
print("WEEK 2 EDA - FINAL SUMMARY")
print("=" * 55)

print("\nDataset dimensions:", df.shape)
print("Duplicate rows:", df.duplicated().sum())

# Count invalid readings in the original data
print("Invalid -200 readings:", (df == -200).sum().sum())

# Summarize numeric columns after replacing invalid readings
numeric_clean = (
    df.select_dtypes(include=np.number)
      .replace(-200, np.nan)
)

print("\nMissing values after invalid-value treatment:")
display(numeric_clean.isna().sum().sort_values(ascending=False))

print("\nDescriptive statistics:")
display(numeric_clean.describe().T.round(3))

print("\nTop positive correlations:")
display(
    correlation_pairs.head(5)
    .to_frame("Correlation")
    .round(3)
)

print("\nTop negative correlations:")
display(
    correlation_pairs.tail(5)
    .sort_values()
    .to_frame("Correlation")
    .round(3)
)

# Export final descriptive statistics
numeric_clean.describe().T.to_csv(
    "week2_final_descriptive_statistics.csv"
)

print("\nFinal descriptive statistics exported successfully.")
```
