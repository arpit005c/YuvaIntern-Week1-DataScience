## 6. Feature Engineering and Data Preparation
### 6.1 Separating Features and Target Variable

```python
# Create feature dataset by removing customerID and Churn
X = data.drop(columns=["customerID", "Churn"])

# Create target variable
y = data["Churn"].map({
    "No": 0,
    "Yes": 1
})

# Display feature and target information
print("Feature Dataset Shape:", X.shape)
print("Target Dataset Shape:", y.shape)

print("\nTarget Variable Distribution:")
print(y.value_counts())

print("\nFeature Columns:")
print(X.columns.tolist())

print("\nTarget Mapping:")
print("No Churn = 0")
print("Churn = 1")
```

### 6.2 Encoding Categorical Features

```python
# Identify categorical and numerical columns
categorical_cols = X.select_dtypes(include=["object"]).columns.tolist()
numerical_cols = X.select_dtypes(exclude=["object"]).columns.tolist()

print("Categorical Columns:", categorical_cols)
print("\nNumerical Columns:", numerical_cols)

# Apply One-Hot Encoding
X_encoded = pd.get_dummies(
    X,
    columns=categorical_cols,
    drop_first=True,
    dtype=int
)

print("\nOriginal Feature Shape:", X.shape)
print("Encoded Feature Shape:", X_encoded.shape)

print("\nFirst 5 Encoded Records:")
display(X_encoded.head())

print("\nMissing Values After Encoding:")
print(X_encoded.isnull().sum().sum())
```

### 6.3 Train-Test Split

```python
from sklearn.model_selection import train_test_split

# Split the encoded features and target
X_train, X_test, y_train, y_test = train_test_split(
    X_encoded,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

# Display dataset sizes
print("Training Features Shape:", X_train.shape)
print("Testing Features Shape:", X_test.shape)

print("\nTraining Target Shape:", y_train.shape)
print("Testing Target Shape:", y_test.shape)

# Check churn distribution in both sets
print("\nTraining Churn Distribution:")
print(y_train.value_counts(normalize=True).mul(100).round(2))

print("\nTesting Churn Distribution:")
print(y_test.value_counts(normalize=True).mul(100).round(2))
```

### 6.4 Feature Scaling

```python
from sklearn.preprocessing import StandardScaler

# Initialize scaler
scaler = StandardScaler()

# Fit on training data and transform
X_train_scaled = scaler.fit_transform(X_train)

# Transform testing data using the same scaler
X_test_scaled = scaler.transform(X_test)

print("Original Training Shape:", X_train.shape)
print("Scaled Training Shape:", X_train_scaled.shape)

print("\nOriginal Testing Shape:", X_test.shape)
print("Scaled Testing Shape:", X_test_scaled.shape)

print("\nFeature scaling completed successfully!")
```

## 7. Machine Learning Model Development
### 7.1 Importing Machine Learning Algorithms

```python
from sklearn.dummy import DummyClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier

# Initialize machine learning models
dummy_model = DummyClassifier(strategy="most_frequent")

logistic_model = LogisticRegression(
    max_iter=1000,
    class_weight="balanced",
    random_state=42
)

random_forest_model = RandomForestClassifier(
    n_estimators=200,
    class_weight="balanced",
    random_state=42,
    n_jobs=-1
)

gradient_model = GradientBoostingClassifier(
    n_estimators=100,
    learning_rate=0.1,
    random_state=42
)

print("Four machine learning models initialized successfully!")
```

```python
# Train the baseline model
dummy_model.fit(X_train, y_train)

# Train Logistic Regression using scaled data
logistic_model.fit(X_train_scaled, y_train)

# Train Random Forest
random_forest_model.fit(X_train, y_train)

# Train Gradient Boosting
gradient_model.fit(X_train, y_train)

print("All four models trained successfully!")
```

```python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score
)

models = {
    "Dummy Classifier": (dummy_model, X_test),
    "Logistic Regression": (logistic_model, X_test_scaled),
    "Random Forest": (random_forest_model, X_test),
    "Gradient Boosting": (gradient_model, X_test)
}

results = []

for name, (model, X_data) in models.items():
    y_pred = model.predict(X_data)
    y_prob = model.predict_proba(X_data)[:, 1]

    results.append({
        "Model": name,
        "Accuracy": accuracy_score(y_test, y_pred),
        "Precision": precision_score(y_test, y_pred, zero_division=0),
        "Recall": recall_score(y_test, y_pred, zero_division=0),
        "F1 Score": f1_score(y_test, y_pred, zero_division=0),
        "ROC-AUC": roc_auc_score(y_test, y_prob)
    })

results_df = pd.DataFrame(results)
results_df = results_df.sort_values(by="F1 Score", ascending=False)

print("Model Performance Comparison:")
display(results_df.round(4))
```

### 7.2 Confusion Matrix Analysis

```python
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.metrics import confusion_matrix

fig, axes = plt.subplots(2, 2, figsize=(12, 9))
axes = axes.flatten()

for ax, (name, (model, X_data)) in zip(axes, models.items()):
    y_pred = model.predict(X_data)
    cm = confusion_matrix(y_test, y_pred)

    sns.heatmap(
        cm,
        annot=True,
        fmt="d",
        cmap="Blues",
        xticklabels=["No Churn", "Churn"],
        yticklabels=["No Churn", "Churn"],
        ax=ax
    )

    ax.set_title(name)
    ax.set_xlabel("Predicted")
    ax.set_ylabel("Actual")

plt.tight_layout()
plt.savefig("../screenshots/07_confusion_matrices.png",
            dpi=300, bbox_inches="tight")
plt.show()
```

### 9. Classification Threshold Optimization
#### 9.1 Finding the Optimal Threshold Using Validation Data

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import precision_recall_curve
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Split training data into training and validation sets
X_train_inner, X_val, y_train_inner, y_val = train_test_split(
    X_train,
    y_train,
    test_size=0.20,
    random_state=42,
    stratify=y_train
)

# Scale using only the inner training data
threshold_scaler = StandardScaler()

X_train_inner_scaled = threshold_scaler.fit_transform(X_train_inner)
X_val_scaled = threshold_scaler.transform(X_val)

# Train a temporary Logistic Regression model
threshold_model = LogisticRegression(
    max_iter=1000,
    class_weight="balanced",
    random_state=42
)

threshold_model.fit(X_train_inner_scaled, y_train_inner)

# Get churn probabilities on validation data
y_val_prob = threshold_model.predict_proba(X_val_scaled)[:, 1]

# Calculate precision and recall at different thresholds
precision, recall, thresholds = precision_recall_curve(
    y_val,
    y_val_prob
)

# Calculate F1-score for each threshold
f1_scores = (
    2 * precision[:-1] * recall[:-1] /
    (precision[:-1] + recall[:-1] + 1e-10)
)

# Find threshold with the highest validation F1-score
best_index = np.argmax(f1_scores)
best_threshold = thresholds[best_index]

print("Threshold Analysis")
print("------------------------------")
print(f"Best Threshold: {best_threshold:.4f}")
print(f"Validation Precision: {precision[best_index]:.4f}")
print(f"Validation Recall: {recall[best_index]:.4f}")
print(f"Validation F1 Score: {f1_scores[best_index]:.4f}")
```

#### 9.2 Threshold Optimization Results

The default classification threshold of 0.50 was adjusted using
a validation dataset.

The optimal threshold was found to be 0.6303 based on the highest
validation F1-score.

Validation results:
- Precision: 0.6070
- Recall: 0.6923
- F1-score: 0.6469

The optimized threshold provides a better balance between
precision and recall on the validation dataset.

The final performance will be evaluated using the untouched
test dataset to ensure an unbiased assessment.

```python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    classification_report
)

# Predict probabilities using the original trained model
y_test_prob = logistic_model.predict_proba(X_test_scaled)[:, 1]

# Apply optimized threshold
y_test_optimized = (
    y_test_prob >= best_threshold
).astype(int)

# Evaluate optimized predictions
print("Optimized Logistic Regression - Test Results")
print("---------------------------------------------")

print("Accuracy:", round(
    accuracy_score(y_test, y_test_optimized), 4
))

print("Precision:", round(
    precision_score(y_test, y_test_optimized), 4
))

print("Recall:", round(
    recall_score(y_test, y_test_optimized), 4
))

print("F1 Score:", round(
    f1_score(y_test, y_test_optimized), 4
))

print("\nConfusion Matrix:")
print(confusion_matrix(y_test, y_test_optimized))

print("\nClassification Report:")
print(classification_report(
    y_test,
    y_test_optimized,
    target_names=["No Churn", "Churn"]
))
```

#### 9.3 Comparison of Default and Optimized Thresholds

The Logistic Regression model was evaluated using both the default
threshold (0.50) and optimized threshold (0.6303).

The optimized threshold achieved:
- Accuracy: 77.15%
- Precision: 55.65%
- Recall: 68.45%
- F1-score: 61.39%

The default threshold achieved:
- Accuracy: 74.02%
- Precision: 50.69%

- F1-score: 61.64%

The optimized threshold improved accuracy and precision but
reduced recall and slightly reduced the F1-score.

Since identifying customers who may churn is the primary project

The optimized threshold experiment demonstrates the trade-off
between precision and recall in binary classification.
